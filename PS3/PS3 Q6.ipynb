## 6a:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('metabric.csv')
```

```python
#parta 
sns.ecdfplot(data =df, x = 'Overall Survival (Months)')
plt.title('ECDF of Overall Survival')
plt.show()
```

## 6b:

```python
#partb
sns.ecdfplot(data =df, x = 'Overall Survival (Months)', hue = 'Chemotherapy')
plt.title('ECDF of Overall Survival Hued by Chemo')
plt.show()
```

Conditional on a patient receiving chemotherapy, we should predict they live a shorter amount of time, due to the orange curve being positioned more to the left than the blue curve, and rising faster. This association doesn't necessarily mean that Chemotherapy is the *cause* of lower survival, though.


## 6c
 We cannot conclude from this analysis whether Chemotherapy is an effective treatment or not. Patients who have undergone Chemotherapy appear to have shorter overall survival, but this could be due to external factors. Most of the time, severely sick people need Chemo, whereas less sick people may not need that intervention - Chemo/Non-Chemo reflect two different sample populations. Scenarios like this aren't accounted for in the ECDF, which is why we cannot make any hard conclusions about effectiveness.
