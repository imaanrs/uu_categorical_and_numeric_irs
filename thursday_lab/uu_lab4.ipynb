```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

## Q1:

```python
#Q1
n = 5_000

for K in [2, 5, 10, 100]:
    grid = []
    for i in range(1, K + 1):
        grid.append(i / K)

    draws = []
    for j in range(n):
        draws.append(np.random.choice(grid))

    draws = np.sort(draws)
    y = np.arange(1, n + 1) / n
    plt.step(draws, y, where='post', label='K = ' + str(K))

plt.plot([0, 1], [0, 1], 'k--', label='straight line')
plt.legend()
plt.show()
```

1.4) As K increases, the ECDF approaches a straight line, which is the CDF of the Uniform distrbution. 

## Q2:

```python
b = 5_000

for n in [1, 2, 5, 32]:
    results = []
    for j in range(b):
        data = np.random.uniform(-np.sqrt(3), np.sqrt(3), n)
        results.append(np.mean(data) * np.sqrt(n))

    results = np.sort(results)
    y = np.arange(1, b + 1) / b
    plt.step(results, y, where='post', label='n = ' + str(n))
    plt.xlabel('x')
    plt.ylabel('F(x)')
x = np.linspace(-4, 4, 200)
plt.plot(x, norm.cdf(x), 'k--', label='standard normal')

plt.legend()
plt.show()
```

2.4) As n increases, the ECDF approaches the CDF of the standard normal distribution. 

## Q3:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

b = 5_000
p = 0.5

for n in [5, 10, 40, 200]:
    results = []
    for j in range(b):
        flips = np.random.rand(n) < p
        heads = np.sum(flips)

        z = (heads - n * p) / np.sqrt(n * p * (1 - p))
        results.append(z)

    results = np.sort(results)
    y = np.arange(1, b + 1) / b
    plt.step(results, y, where='post', label='n = ' + str(n))
    plt.xlabel('x')
    plt.ylabel('F(x)')
    
x = np.linspace(-4, 4, 200)
plt.plot(x, norm.cdf(x), 'k--', label='standard normal')
plt.legend()
plt.show()
```

3.5) As n increases, the ECDF approaches the standard normal distribution.

3.6) When p=0.5, the results are symmetric & the ECDF approaches the standard normal distr. quickly. When p moves towards 0 or 1, the results become skewed and the ECDF shows a few big jumps instead of the smooth S shape. It still appraoches the normal as n grows but the further p is from 0.5 the larger n needs to be. 

## Q4:

```python
n = 5_000
r = 1

for dt in [0.5, 0.1, 0.01, 0.001]:
    results = []
    for j in range(n):
        period = 0
        while True:
            period = period + 1
            if np.random.rand() < r * dt:
                break

        results.append(period * dt)

    results = np.sort(results)
    y = np.arange(1, n + 1) / n
    plt.step(results, y, where='post', label='dt = ' + str(dt))
    plt.xlabel('x')
    plt.ylabel('F(x)')
    
x = np.linspace(0, 6, 200)
plt.plot(x, 1 - np.exp(-r * x), 'k--', label='exponential')
plt.legend()
plt.show()
```

```python
n = 5_000
r = 2

for dt in [0.25, 0.1, 0.01, 0.001]:
    results = []
    for j in range(n):
        period = 0
        while True:
            period = period + 1
            if np.random.rand() < r * dt:
                break

        results.append(period * dt)

    results = np.sort(results)
    y = np.arange(1, n + 1) / n
    plt.step(results, y, where='post', label='dt = ' + str(dt))
    plt.xlabel('x')
    plt.ylabel('F(x)')
    
x = np.linspace(0, 3, 200)
plt.plot(x, 1 - np.exp(-r * x), 'k--', label='exponential')
plt.legend()
plt.show()
```

4.4) As dt goes to 0, the ECDF approaches the exponential distribution. AT dt = 0.5, we see big steps and at dt = 0.001 the staircase is pretty much one with the dashed line.

4.5) Increasing r raises the chance of termination in each period so it happens sooner. The ECDF rises steeply & the mean time 1/r gets shorter. r only chnages the time scale so the shape is still exponential.
